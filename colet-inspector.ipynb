```python
import os

import numpy as np

import pandas as pd

from sklearn.model_selection import train_test_split
```

```python
colet_path = '/work/cvcs2026/LZMM/COLET/dumped/'

gaze_df = pd.read_csv(os.path.join(colet_path, 'gaze_dataset.csv'), sep=',')
pupil_df = pd.read_csv(os.path.join(colet_path, 'pupil_dataset.csv'), sep=',')
blinks_df = pd.read_csv(os.path.join(colet_path, 'blinks_dataset.csv'), sep=',')
annotation_df = pd.read_csv(os.path.join(colet_path, 'annotation_dataset.csv'), sep=',')    # ground truth
```

```python
metrics = ['mental', 'physical', 'temporal', 'performance', 'effort', 'frustration', 'mean']

for m in metrics:
    annotation_df[m] = annotation_df.apply(
        lambda row: row[f"{m}_{int(row['task_number'])}"], 
        axis=1
    )

cols_to_drop = [f"{m}_{i}" for m in metrics for i in range(1, 5)]
annotation_df = annotation_df.drop(columns=cols_to_drop)
```

```python
print('Gaze:')
print(gaze_df.columns.to_list())
display(gaze_df.head(3))

print('Pupil:')
print(pupil_df.columns.to_list())
display(pupil_df.head(3))

print('Blinks:')
print(blinks_df.columns.to_list())
display(blinks_df.head(3))

print('Annotation:')
print(annotation_df.columns.to_list())
display(annotation_df.head(4))
```

```python
gaze_drop = ['Education', 'world_index', 'norm_pos_x', 'norm_pos_y', 'base_data', 'gaze_point_3d_x', 'gaze_point_3d_y', 'gaze_point_3d_z', 'eye_center0_3d_x', 'eye_center0_3d_y', 'eye_center0_3d_z', 'gaze_normal0_x', 'gaze_normal0_y', 'gaze_normal0_z', 'eye_center1_3d_x', 'eye_center1_3d_y', 'eye_center1_3d_z', 'gaze_normal1_x', 'gaze_normal1_y', 'gaze_normal1_z']
gaze_df_reduced = gaze_df.drop(columns=gaze_drop)

pupil_drop = ['VisualAcuity_logMAR_', 'confidence', 'Education', 'world_index', 'norm_pos_x', 'norm_pos_y', 'method', 'ellipse_center_x', 'ellipse_center_y', 'ellipse_axis_a', 'ellipse_axis_b', 'ellipse_angle', 'diameter_3d', 'model_confidence', 'model_id', 'sphere_center_x', 'sphere_center_y', 'sphere_center_z', 'sphere_radius', 'circle_3d_center_x', 'circle_3d_center_y', 'circle_3d_center_z', 'circle_3d_normal_x', 'circle_3d_normal_y', 'circle_3d_normal_z', 'circle_3d_radius', 'theta', 'phi', 'projected_sphere_center_x', 'projected_sphere_center_y', 'projected_sphere_axis_a', 'projected_sphere_axis_b', 'projected_sphere_angle']
pupil_df_reduced = pupil_df.drop(columns=pupil_drop)

blinks_drop = ['Education', 'id', 'end_timestamp', 'start_frame_index', 'index', 'end_frame_index', 'filter_response', 'base_data']
blinks_df_reduced = blinks_df.drop(columns=blinks_drop)

gaze_df_reduced = gaze_df_reduced.sort_values(['subject_id', 'task_number', 'gaze_timestamp'], ascending=True)
pupil_df_reduced = pupil_df_reduced.sort_values(['subject_id', 'task_number', 'pupil_timestamp'], ascending=True)
blinks_df_reduced = blinks_df_reduced.sort_values(['subject_id', 'task_number', 'start_timestamp'], ascending=True)

print('Gaze:')
display(gaze_df_reduced.dtypes)
display(gaze_df_reduced.head(3))
print('Pupil:')
display(pupil_df_reduced.dtypes)
display(pupil_df_reduced.head(3))
print('Blinks:')
display(blinks_df_reduced.dtypes)
display(blinks_df_reduced.head(3))
```

```python
label = annotation_df.drop(columns=['VisualAcuity_logMAR_', 'Gender', 'Age', 'Education', 'physical', 'temporal', 'performance', 'effort', 'frustration', 'mean'])
label['cload'] = (label['mental'] >= 50).astype(int)
label = label.drop(columns=['mental'])
label.head(2)
```

```python
pupil_df_reduced = pupil_df_reduced.merge(label, on=['subject_id', 'task_number'])
pupil_df_reduced.head(1)
```

```python
pupil_activities = []

for subj in range(1, 48):
    for task in range(1, 5):
        for eye in range(2):
            pupil_activity = pupil_df_reduced[(pupil_df_reduced.subject_id == subj) & (pupil_df_reduced.task_number == task) & (pupil_df_reduced.eye_id == eye)].reset_index(drop=True)
            pupil_activity = pupil_activity.sort_values('pupil_timestamp')
            pupil_activities.append(pupil_activity)
print('Activities number:', len(pupil_activities))

for pupil_activity in pupil_activities[:4]:
    display(pupil_activity.head(1))
    print(len(pupil_activity))
```
